```
print("=== Program Perhitungan Daya Sepeda")

# input data
m = float(input("Massa (kg)         : "))
v = float(input("Kecepatan (m/s)    : "))
t = float(input("Sudut (derajat)    : "))

g = 9.8

# perhitungan
import math
P = 2 * m * g * math.sin(math.radians(t)) * v

print("\n==== HASIL ====")
print("Daya :", round(P, 2), "Watt")
print("Ilma Azizah")
print("K1C025128")
```
