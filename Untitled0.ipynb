```python
print("=== Program Wheatstone Bridge ===")

R1 = float(input("Masukkan R1: "))
R2 = float(input("Masukkan R2: "))
R3 = float(input("Masukkan R3: "))

Rx = (R2 * R3) / R1

#Output
print("Nilai Rx =", round(Rx, 2), "Ohm")

```
